### 1.2 NumPy Operations

```python
# Create a NumPy array
sales = np.array([125, 138, 142, 155, 168, 180, 195, 210, 225, 240, 258, 275])

print(f"Original Sales Array: {sales}")

# Display array properties
print(f"\nArray Properties:")
print(f"Shape: {sales.shape}")        # Dimensions of the array
print(f"Size: {sales.size}")          # Total number of elements
print(f"Data Type (dtype): {sales.dtype}") # Data type of elements

# Display basic statistical measures
print(f"\nStatistical Measures:")
print(f"Sum: {sales.sum()}")
print(f"Mean: {sales.mean():.2f}")
print(f"Maximum: {sales.max()}")
print(f"Minimum: {sales.min()}")
print(f"Standard Deviation: {sales.std():.2f}")
```

```python
# Reshape the array into a 4x3 matrix
sales_reshaped = sales.reshape(4, 3)
print(f"\nReshaped Sales Array (4x3):\n{sales_reshaped}")
```

```python
# Find indices where sales are greater than 200
indices_gt_200 = np.where(sales > 200)
print(f"\nIndices where sales > 200: {indices_gt_200[0]}")
print(f"Sales values where sales > 200: {sales[indices_gt_200]}")
```

```python
# Calculate month-over-month growth using np.diff()
# np.diff() computes the difference between consecutive elements
month_over_month_growth = np.diff(sales)
print(f"\nMonth-over-month Growth (using np.diff()): {month_over_month_growth}")
```

## Task 2: Pandas DataFrames

### 2.1 Create DataFrames

```python
# Create Customers DataFrame
customers_data = {
    'cust_id': ['C01', 'C02', 'C03', 'C04', 'C05', 'C06', 'C07', 'C08'],
    'name': ['Aarav', 'Bhavya', 'Chirag', 'Diya', 'Eshan', 'Farah', 'Gaurav', 'Hina'],
    'city': ['Mumbai', 'Delhi', 'Bangalore', 'Mumbai', 'Pune', 'Delhi', 'Bangalore', 'Mumbai'],
    'age': [28, 32, 26, np.nan, 30, 29, 40, 27],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F', 'M', np.nan]
}
customers_df = pd.DataFrame(customers_data)

# Create Purchases DataFrame
purchases_data = {
    'purchase_id': ['P101', 'P102', 'P103', 'P104', 'P105', 'P106', 'P107', 'P108', 'P109', 'P110'],
    'cust_id': ['C01', 'C02', 'C03', 'C01', 'C05', 'C02', 'C06', 'C07', 'C03', 'C05'],
    'product': ['Laptop', 'Saree', 'Smartphone', 'Tablet', 'Headphones', 'Watch', 'T-shirt', 'Backpack', 'Mouse', 'Charger'],
    'amount': [55000, 4500, 32000, 18000, 1500, 3500, 499, 1800, 899, 650],
    'purchase_date': ['2025-02-03', '2025-02-05', '2025-02-09', '2025-02-11', '2025-02-12', '2025-02-14', '2025-02-18', '2025-02-20', '2025-02-22', '2025-02-26']
}
purchases_df = pd.DataFrame(purchases_data)

print("Customers DataFrame (first 5 rows):")
display(customers_df.head())

print("\nPurchases DataFrame (first 5 rows):")
display(purchases_df.head())
```

### 2.2 Data Information and Handling Missing Values

```python
print("\nCustomers DataFrame Info:")
customers_df.info()

print("\nPurchases DataFrame Info:")
purchases_df.info()
```

```python
print("\nMissing values in Customers DataFrame:")
display(customers_df.isnull().sum())

# Fill missing 'age' with the mean age
mean_age = customers_df['age'].mean()
customers_df['age'].fillna(mean_age, inplace=True)

# Fill missing 'gender' with "Unknown"
customers_df['gender'].fillna('Unknown', inplace=True)

print("\nCustomers DataFrame after filling missing values:")
display(customers_df.head())

print("\nMissing values in Customers DataFrame after handling:")
display(customers_df.isnull().sum())
```

```python
# Convert 'age' to integer type
customers_df['age'] = customers_df['age'].astype(int)

# Convert 'purchase_date' to datetime objects in purchases_df
purchases_df['purchase_date'] = pd.to_datetime(purchases_df['purchase_date'])

print("\nCustomers DataFrame Info after type conversion:")
customers_df.info()

print("\nPurchases DataFrame Info after type conversion:")
purchases_df.info()
```

### 2.3 Merging DataFrames

```python
# Merge customers_df and purchases_df on 'cust_id'
merged_df = pd.merge(customers_df, purchases_df, on='cust_id', how='inner')

print("Merged DataFrame (first 5 rows):")
display(merged_df.head())

print("\nShape of Merged DataFrame:")
print(merged_df.shape)
```

## Task 3: Data Filtering and Grouping

### 3.1 Filtering Data

```python
# Show Mumbai purchases
mumbai_purchases = merged_df[merged_df['city'] == 'Mumbai']
print("Purchases made by customers from Mumbai:")
display(mumbai_purchases)
```

```python
# Show purchases where amount > 5000
high_value_purchases = merged_df[merged_df['amount'] > 5000]
print("Purchases with amount greater than 5000:")
display(high_value_purchases)
```

```python
# Show male customers who made purchases
male_customers_purchases = merged_df[merged_df['gender'] == 'M']
print("Purchases made by male customers:")
display(male_customers_purchases)
```

### 3.2 Grouping Data

```python
# Total purchase amount per city
total_amount_per_city = merged_df.groupby('city')['amount'].sum().reset_index()
total_amount_per_city = total_amount_per_city.rename(columns={'amount': 'Total Purchase Amount'})
print("Total purchase amount per city:")
display(total_amount_per_city)
```

```python
# Average purchase amount per gender
average_amount_per_gender = merged_df.groupby('gender')['amount'].mean().reset_index()
average_amount_per_gender = average_amount_per_gender.rename(columns={'amount': 'Average Purchase Amount'})
print("Average purchase amount per gender:")
display(average_amount_per_gender)
```

```python
# Number of purchases per customer
purchases_per_customer = merged_df.groupby(['cust_id', 'name'])['purchase_id'].count().reset_index()
purchases_per_customer = purchases_per_customer.rename(columns={'purchase_id': 'Number of Purchases'})
print("Number of purchases per customer:")
display(purchases_per_customer)
```

```python
# Top 3 customers by total spending
top_3_customers = merged_df.groupby(['cust_id', 'name'])['amount'].sum().nlargest(3).reset_index()
top_3_customers = top_3_customers.rename(columns={'amount': 'Total Spending'})
print("Top 3 customers by total spending:")
display(top_3_customers)
```

## Task 4: Visualization

### 4.1 Bar Chart: Total Purchase Amount by City

```python
# Calculate total purchase amount per city
total_amount_per_city = merged_df.groupby('city')['amount'].sum().reset_index()

# Create a bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='city', y='amount', data=total_amount_per_city, palette='viridis')
plt.title('Total Purchase Amount by City')
plt.xlabel('City')
plt.ylabel('Total Purchase Amount')
plt.show()
```

### 4.2 Pie Chart: Share of Purchases by Gender

```python
# Calculate total purchase amount per gender
total_amount_per_gender = merged_df.groupby('gender')['amount'].sum()

# Create a pie chart
plt.figure(figsize=(8, 8))
plt.pie(total_amount_per_gender, labels=total_amount_per_gender.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Share of Purchases by Gender')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

### 4.3 Bar Chart: Top 3 Customers by Total Spending

```python
# Calculate total spending per customer
total_spending_per_customer = merged_df.groupby('name')['amount'].sum().nlargest(3).reset_index()

# Create a bar chart for top 3 customers
plt.figure(figsize=(10, 6))
sns.barplot(x='name', y='amount', data=total_spending_per_customer, palette='magma')
plt.title('Top 3 Customers by Total Spending')
plt.xlabel('Customer Name')
plt.ylabel('Total Spending')
plt.show()
```

### 4.4 Line Chart: Purchase Date vs Amount

```python
# Sort data by purchase date for a clear line plot
purchases_by_date = merged_df.sort_values(by='purchase_date')

# Create a line chart
plt.figure(figsize=(12, 7))
sns.lineplot(x='purchase_date', y='amount', data=purchases_by_date, marker='o', color='green')
plt.title('Purchase Amount Over Time')
plt.xlabel('Purchase Date')
plt.ylabel('Purchase Amount')
plt.xticks(rotation=45)
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```
